# Check Connection Layer

```python
from bitpin_utils import fetch_data

print("⚡ Connection Layer Test")
print("wallet details")

wallet_data = fetch_data("/api/v1/wlt/wallets/")
i = 0
for item in wallet_data:
    if float(item["balance"]) > 0:
        i += 1
        print("{:02d})  {:10}  balance: {:10}  frozen: {:10}".format(i, item["asset"], item["balance"], item["frozen"]))
if i == 0:
    print("Wallet is empty.")
```
